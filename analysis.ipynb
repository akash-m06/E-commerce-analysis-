# E-Commerce Sales Analytics — Exploratory Data AnalysisThis notebook explores a synthetic e-commerce transactions dataset covering 2024–2025:revenue trends, category/product performance, regional breakdown, customer segments,payment methods, and return rates.Data source: `data/ecommerce_data.csv` (generated by `data/generate_data.py`).

```python
import pandas as pdimport matplotlib.pyplot as pltimport seaborn as snssns.set_style("whitegrid")plt.rcParams["figure.figsize"] = (10, 6)df = pd.read_csv("../data/ecommerce_data.csv", parse_dates=["order_date"])df["month"] = df["order_date"].dt.to_period("M").astype(str)df.head()
```

## 1. Dataset Overview

```python
completed = df[df["order_status"] == "Completed"].copy()print(f"Total orders: {len(df):,}")print(f"Date range: {df['order_date'].min().date()} to {df['order_date'].max().date()}")print(f"Unique customers: {df['customer_id'].nunique():,}")print(f"Completed orders: {len(completed):,} ({len(completed)/len(df):.1%})")print(f"Total revenue (completed): ${completed['net_amount'].sum():,.2f}")print(f"Average order value: ${completed['net_amount'].mean():,.2f}")
```

## 2. Monthly Revenue Trend

```python
monthly_revenue = completed.groupby("month")["net_amount"].sum().reset_index()plt.figure()plt.plot(monthly_revenue["month"], monthly_revenue["net_amount"], marker="o", color="#2563eb")plt.xticks(rotation=45, ha="right")plt.title("Monthly Revenue Trend")plt.ylabel("Revenue ($)")plt.tight_layout()plt.show()
```

## 3. Revenue by Category

```python
category_revenue = completed.groupby("category")["net_amount"].sum().sort_values(ascending=False)plt.figure()sns.barplot(x=category_revenue.values, y=category_revenue.index, hue=category_revenue.index, palette="Blues_r", legend=False)plt.title("Revenue by Category")plt.xlabel("Revenue ($)")plt.tight_layout()plt.show()
```

## 4. Top 10 Products by Revenue

```python
top_products = completed.groupby("product")["net_amount"].sum().sort_values(ascending=False).head(10)plt.figure()sns.barplot(x=top_products.values, y=top_products.index, hue=top_products.index, palette="Greens_r", legend=False)plt.title("Top 10 Products by Revenue")plt.xlabel("Revenue ($)")plt.tight_layout()plt.show()
```

## 5. Revenue by Region

```python
region_revenue = completed.groupby("region")["net_amount"].sum().sort_values(ascending=False)plt.figure()sns.barplot(x=region_revenue.index, y=region_revenue.values, hue=region_revenue.index, palette="Oranges_r", legend=False)plt.title("Revenue by Region")plt.ylabel("Revenue ($)")plt.tight_layout()plt.show()
```

## 6. Customer Segment Performance

```python
segment_summary = completed.groupby("customer_segment").agg(    orders=("order_id", "count"),    revenue=("net_amount", "sum"),    avg_order_value=("net_amount", "mean"),).round(2).sort_values("revenue", ascending=False)segment_summary
```

```python
plt.figure()plt.pie(    segment_summary["revenue"],    labels=segment_summary.index,    autopct="%1.1f%%",    colors=["#60a5fa", "#34d399", "#fbbf24"],)plt.title("Revenue Share by Customer Segment")plt.tight_layout()plt.show()
```

## 7. Order Status & Returns

```python
status_counts = df["order_status"].value_counts(normalize=True) * 100status_counts.round(2)
```

## 8. Payment Method Popularity

```python
payment_counts = df["payment_method"].value_counts()plt.figure()sns.barplot(x=payment_counts.values, y=payment_counts.index, hue=payment_counts.index, palette="Purples_r", legend=False)plt.title("Orders by Payment Method")plt.xlabel("Number of Orders")plt.tight_layout()plt.show()
```

## Key Takeaways- Revenue peaks in October–November, aligned with festive-season discounting.- Electronics and Fashion are the top two revenue-generating categories.- VIP customers have the highest average order value despite being the smallest segment.- UPI and Credit Card are the dominant payment methods.- Return + cancellation rate sits around 7%, with variation by category (see `sql/analysis_queries.sql`, query 7).
